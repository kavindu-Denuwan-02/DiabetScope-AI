```python
# Environment Setup & Imports
# ─────────────────────────────────────────────────────
# 1. IMPORTS
# ─────────────────────────────────────────────────────
# pandas   → data loading, manipulation, summary stats
# numpy    → numerical operations, array handling
# matplotlib / seaborn → visualisations
# warnings → suppress non-critical output noise

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')

# ─────────────────────────────────────────────────────
# 2. PLOT STYLE
# ─────────────────────────────────────────────────────
# Set a clean, consistent visual style for every chart.
# 'ggplot' gives a white-grid look that reads well in notebooks.
# figsize default keeps every plot uniformly sized.

plt.style.use('ggplot')
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['axes.titlesize']  = 14
plt.rcParams['axes.labelsize']  = 12

print('✅ All imports successful')
```

```python

# We NEVER overwrite this file. All cleaning output
# goes to data/processed/.
# ─────────────────────────────────────────────────────

RAW_PATH = '../data/raw/diabetes.csv'

df = pd.read_csv(RAW_PATH)

print(f'Shape : {df.shape[0]} rows × {df.shape[1]} columns')
print(f'Memory: {df.memory_usage(deep=True).sum() / 1024:.1f} KB')
df.head(10)  # first 10 rows — quick sanity check
```

```python

# Pregnancies                → number of times pregnant
# Glucose                    → 2-hour plasma glucose (oral glucose tolerance test)
# BloodPressure              → diastolic blood pressure (mm Hg)
# SkinThickness              → triceps skin fold thickness (mm)
# Insulin                    → 2-hour serum insulin (μU/mL)
# BMI                        → Body Mass Index (kg/m²)
# DiabetesPedigreeFunction   → genetic likelihood score (0.078 – 2.42 range)
# Age                        → age in years (min 21 by dataset constraint)
# Outcome                    → TARGET: 1 = diabetic, 0 = non-diabetic

column_info = {
    'Column': df.columns.tolist(),
    'Dtype': df.dtypes.tolist(),
    'Non-Null': df.notna().sum().tolist(),
    'Unique': df.nunique().tolist()
}

info_df = pd.DataFrame(column_info)
print(info_df.to_string(index=False))
```

```python
# ─────────────────────────────────────────────────────
# describe() gives: count, mean, std, min, 25%, 50%, 75%, max
# .T  → transpose so each column is a row (easier to read)
#
# KEY OBSERVATIONS TO LOOK FOR:
#   • min = 0 where 0 is biologically impossible (→ missing values)
#   • Large std relative to mean (→ high variance / potential outliers)
#   • Skew: if median << mean, the distribution is right-skewed
# ─────────────────────────────────────────────────────

df.describe().T
```

```python
# ─────────────────────────────────────────────────────
# CRITICAL INSIGHT:
# This dataset has NO actual NaN values.
# However, several columns use 0 as a placeholder for
# "measurement not taken" — which is biologically impossible:
#
#   Glucose           →  0 mg/dL is not survivable        → 5 zeros  (0.65%)
#   BloodPressure     →  0 mm Hg means no heartbeat       → 35 zeros (4.56%)
#   SkinThickness     →  0 mm is anatomically impossible   → 227 zeros (29.56%)
#   Insulin           →  0 μU/mL not possible in living   → 374 zeros (48.70%)
#   BMI               →  0 kg/m² is not possible           → 11 zeros (1.43%)
#
# Pregnancies = 0 is VALID (never been pregnant).
# Age = 0 does NOT appear (min is 21).
# DiabetesPedigreeFunction = 0 does NOT appear (min is 0.078).
# ─────────────────────────────────────────────────────

# Columns where 0 is a "fake" value (biologically impossible)
ZERO_MEANING_MISSING = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

zero_report = pd.DataFrame({
    'Feature': ZERO_MEANING_MISSING,
    'Zero Count': [( df[c] == 0).sum() for c in ZERO_MEANING_MISSING],
    'Zero %': [(df[c] == 0).mean() * 100 for c in ZERO_MEANING_MISSING]
})

print(zero_report.to_string(index=False))
print(f"\nTotal rows with ≥1 fake zero: {(df[ZERO_MEANING_MISSING].eq(0).any(axis=1)).sum()} / {len(df)} "
      f"({(df[ZERO_MEANING_MISSING].eq(0).any(axis=1)).mean()*100:.1f}%)")
```

```python
# ─────────────────────────────────────────────────────
# VISUALISE the zero distribution:
# A horizontal bar chart comparing zero-count severity.
# Insulin and SkinThickness are the biggest problems.
# ─────────────────────────────────────────────────────

fig, ax = plt.subplots(figsize=(9, 4))

colors = ['#e74c3c' if p > 20 else '#f39c12' if p > 5 else '#2ecc71'
          for p in zero_report['Zero %']]

ax.barh(zero_report['Feature'], zero_report['Zero %'], color=colors, edgecolor='black', height=0.5)
ax.set_xlabel('Percentage of Zeros (%)')
ax.set_title('Zero-Value Distribution (Fake Missings per Feature)')
ax.axvline(x=5, color='gray', linestyle='--', label='5% threshold')

# Label each bar with the exact percentage
for i, v in enumerate(zero_report['Zero %']):
    ax.text(v + 0.3, i, f'{v:.1f}%', va='center', fontsize=11)

ax.legend()
plt.tight_layout()
plt.show()
```

```python
# ─────────────────────────────────────────────────────
# TARGET: Outcome
#   0 → Non-diabetic : 500 samples (65.1%)
#   1 → Diabetic     : 268 samples (34.9%)
#
# IMBALANCE RATIO ≈ 1.87 : 1  (majority : minority)
#
# This is a MODERATE imbalance.
# • Not extreme enough to break most algorithms outright
# • But significant enough that accuracy alone is misleading
#   (a naive "always predict 0" model would score 65.1% accuracy)
# • We MUST use stratified splits & consider resampling later.
# ─────────────────────────────────────────────────────

target_counts = df['Outcome'].value_counts().sort_index()
target_pcts   = df['Outcome'].value_counts(normalize=True).sort_index() * 100

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# --- Left: Count bar chart ---
axes[0].bar(['0 (Non-Diabetic)', '1 (Diabetic)'],
            target_counts.values,
            color=['#3498db', '#e74c3c'], edgecolor='black', width=0.5)
axes[0].set_title('Target Distribution — Counts')
axes[0].set_ylabel('Count')
for i, v in enumerate(target_counts.values):
    axes[0].text(i, v + 5, str(v), ha='center', fontsize=13, fontweight='bold')

# --- Right: Pie chart ---
axes[1].pie(target_counts.values,
            labels=['Non-Diabetic (65.1%)', 'Diabetic (34.9%)'],
            colors=['#3498db', '#e74c3c'],
            autopct='',
            startangle=90,
            explode=(0.03, 0.03))
axes[1].set_title('Target Distribution — Proportion')

plt.tight_layout()
plt.show()

print(f"\nImbalance ratio (Non-Diabetic / Diabetic): {target_counts[0]/target_counts[1]:.2f} : 1")
```

```python
# ─────────────────────────────────────────────────────
# WHY histograms + KDE overlay?
# • Histogram shows the raw frequency bins
# • KDE (Kernel Density Estimate) smooths it into a continuous
#   curve so we can see the overall shape clearly
#
# COLOR CODING:
#   Blue  = Non-diabetic (Outcome 0)
#   Red   = Diabetic     (Outcome 1)
#
# This lets us see WHERE in each feature's range diabetic
# patients tend to cluster vs non-diabetic.
# ─────────────────────────────────────────────────────

features = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness',
            'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age']

fig, axes = plt.subplots(2, 4, figsize=(20, 9))
axes = axes.flatten()

for i, col in enumerate(features):
    ax = axes[i]
    
    # Plot each class separately
    for outcome, color, label in [(0, '#3498db', 'Non-Diabetic'), (1, '#e74c3c', 'Diabetic')]:
        subset = df.loc[df['Outcome'] == outcome, col]
        ax.hist(subset, bins=30, alpha=0.45, color=color, label=label, edgecolor='black', linewidth=0.5)
        # KDE line
        subset.plot.kde(ax=ax, color=color, linewidth=2)
    
    ax.set_title(col)
    ax.set_xlabel('Value')
    ax.set_ylabel('Frequency')
    ax.legend(fontsize=8)

plt.suptitle('Feature Distributions by Diabetes Status (Histogram + KDE)', y=1.02, fontsize=15)
plt.tight_layout()
plt.show()
```

```python
# ─────────────────────────────────────────────────────
# BOX PLOTS — outlier detection
#
# The box shows Q1–Q3 (middle 50% of data).
# The whiskers extend to 1.5×IQR.
# Dots beyond the whiskers = statistical outliers.
#
# Pre-calculated outlier counts from profiling:
#   Pregnancies              →  4 outliers  (values > 13.5)
#   BloodPressure            → 45 outliers  (values > 107)
#   Insulin                  → 34 outliers  (values > 318)
#   DiabetesPedigreeFunction → 29 outliers  (values > 1.2)
#   BMI                      → 19 outliers  (values > 50.55)
#   Age                      →  9 outliers  (values > 66.5)
#   SkinThickness            →  1 outlier
#   Glucose                  →  5 outliers  (but those are the zeros!)
# ─────────────────────────────────────────────────────

fig, axes = plt.subplots(2, 4, figsize=(20, 8))
axes = axes.flatten()

for i, col in enumerate(features):
    ax = axes[i]
    
    # Color by class so we can see which class has outliers
    data_0 = df.loc[df['Outcome'] == 0, col]
    data_1 = df.loc[df['Outcome'] == 1, col]
    
    bp = ax.boxplot([data_0, data_1],
                    labels=['Non-Diabetic', 'Diabetic'],
                    patch_artist=True,
                    widths=0.5)
    
    # Color the boxes
    bp['boxes'][0].set_facecolor('#3498db')
    bp['boxes'][1].set_facecolor('#e74c3c')
    
    ax.set_title(col)
    ax.tick_params(axis='x', labelsize=8)

plt.suptitle('Boxplots by Diabetes Status — Outlier Inspection', y=1.02, fontsize=15)
plt.tight_layout()
plt.show()
```

```python
# ─────────────────────────────────────────────────────
# WHY correlation analysis?
# • Identifies which features move together (multicollinearity)
# • Shows which features have the strongest LINEAR signal with Outcome
#
# Pre-computed correlations with Outcome (sorted):
#   Glucose                   0.467  ← STRONGEST signal by far
#   BMI                       0.293  ← Second strongest
#   Age                       0.238
#   Pregnancies               0.222
#   DiabetesPedigreeFunction  0.174
#   Insulin                   0.131
#   SkinThickness             0.075  ← Very weak
#   BloodPressure             0.065  ← Weakest
#
# Note: Low correlation ≠ useless.
# Tree-based models can capture non-linear relationships
# that Pearson correlation misses.
# ─────────────────────────────────────────────────────

corr_matrix = df.corr()

fig, ax = plt.subplots(figsize=(10, 8))

# Mask the upper triangle to avoid redundant info
mask = np.triu(np.ones_like(corr_matrix, dtype=bool), k=1)

sns.heatmap(
    corr_matrix,
    mask=mask,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0,
    vmin=-1, vmax=1,
    linewidths=0.5,
    cbar_kws={'shrink': 0.8},
    ax=ax
)

ax.set_title('Correlation Heatmap (Lower Triangle)', pad=15)
plt.tight_layout()
plt.show()
```

```python
# ─────────────────────────────────────────────────────
# FOCUSED: Correlation with target only
# Horizontal bar chart, sorted by absolute value.
# This is the single most useful chart for feature importance intuition.
# ─────────────────────────────────────────────────────

target_corr = corr_matrix['Outcome'].drop('Outcome').sort_values(ascending=True)

fig, ax = plt.subplots(figsize=(9, 5))

colors = ['#e74c3c' if v > 0.2 else '#f39c12' if v > 0.1 else '#95a5a6'
          for v in target_corr.values]

ax.barh(target_corr.index, target_corr.values, color=colors, edgecolor='black', height=0.5)
ax.axvline(x=0, color='black', linewidth=0.8)
ax.axvline(x=0.2, color='red', linestyle='--', alpha=0.5, label='0.2 threshold')
ax.set_xlabel('Pearson Correlation with Outcome')
ax.set_title('Feature Correlation with Target (Outcome)')
ax.legend()

for i, v in enumerate(target_corr.values):
    ax.text(v + 0.005, i, f'{v:.3f}', va='center', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
# ─────────────────────────────────────────────────────
# IQR METHOD explained:
#   IQR = Q3 - Q1  (the interquartile range)
#   Lower fence = Q1 - 1.5 × IQR
#   Upper fence = Q3 + 1.5 × IQR
#   Any value outside [lower_fence, upper_fence] → outlier
#
# We compute this NOW so the preprocessing notebook
# can reference the exact cap values.
# ─────────────────────────────────────────────────────

outlier_report = []

for col in features:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    n_outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    pct = n_outliers / len(df) * 100
    
    outlier_report.append({
        'Feature': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': round(IQR, 3),
        'Lower Fence': round(lower, 3),
        'Upper Fence': round(upper, 3),
        'Outlier Count': n_outliers,
        'Outlier %': round(pct, 2)
    })

outlier_df = pd.DataFrame(outlier_report)
print(outlier_df.to_string(index=False))
print(f"\n📌 These fence values will be used as CAP limits in Notebook 02.")
```
