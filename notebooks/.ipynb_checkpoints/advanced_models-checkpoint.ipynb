```python
# ─────────────────────────────────────────────────────
# IMPORTS:
#   RandomForestClassifier  → ensemble of decision trees
#   XGBClassifier           → gradient boosted trees (pip install xgboost)
#   GridSearchCV            → exhaustive hyperparameter search
#   StratifiedKFold         → ensures each CV fold keeps class balance
# ─────────────────────────────────────────────────────

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json
import os
import warnings

from sklearn.ensemble       import RandomForestClassifier
from xgboost                import XGBClassifier
from sklearn.model_selection import GridSearchCV, StratifiedKFold, train_test_split
from sklearn.preprocessing  import StandardScaler
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, matthews_corrcoef, roc_auc_score,
    roc_curve, confusion_matrix, classification_report
)
import joblib

warnings.filterwarnings('ignore')
plt.style.use('ggplot')
plt.rcParams['figure.figsize'] = (10, 6)

# ─────────────────────────────────────────────────────
# Reload cleaned data & recreate the exact same split
# Same seed = same train/test rows every single time
# ─────────────────────────────────────────────────────

df = pd.read_csv('../data/processed/diabetes_clean.csv')
X  = df.drop(columns=['Outcome'])
y  = df['Outcome']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# Scale (fit on train only)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled  = scaler.transform(X_test)

print(f'Data ready → Train: {X_train_scaled.shape} | Test: {X_test_scaled.shape}')
```

```python
# ─────────────────────────────────────────────────────
# Same evaluation function as Notebook 03.
# Keeps metrics consistent across all 5 models.
# ─────────────────────────────────────────────────────

def evaluate_model(model_name, y_true, y_pred, y_prob):
    return {
        'Model'    : model_name,
        'Recall'   : recall_score(y_true, y_pred, pos_label=1),
        'F1-Score' : f1_score(y_true, y_pred, pos_label=1),
        'MCC (Phi)': matthews_corrcoef(y_true, y_pred),
        'ROC-AUC'  : roc_auc_score(y_true, y_prob),
        'Accuracy' : accuracy_score(y_true, y_pred),
        'Precision': precision_score(y_true, y_pred, pos_label=1)
    }

all_results = []  # collects results from both models

print('✅ Evaluation function ready')
```

```python
# ─────────────────────────────────────────────────────
# RANDOM FOREST — an ensemble of many decision trees
#
# What it does:
#   1. Bootstrap: randomly sample rows WITH replacement
#   2. For each tree, randomly pick a SUBSET of features
#   3. Train a full decision tree on that subset
#   4. Repeat N times (n_estimators)
#   5. Final prediction = majority vote across all trees
#
# Why it works well here:
#   • Each tree sees different data + different features
#     → trees are diverse → ensemble is robust
#   • Handles feature interactions automatically
#   • Resistant to overfitting (unlike a single tree)
#   • Does NOT need scaled data (but we pass it anyway
#     for consistency; it won't hurt)
#
# GRIDSEARCHCV explained:
#   Tries EVERY combination of the parameter grid.
#   For each combo, runs 5-fold cross-validation.
#   Picks the combo with the best CV score.
#
# WHY scoring='recall'?
#   Our #1 medical priority is catching diabetic patients.
#   Recall = TP / (TP + FN) = fraction of actual diabetics caught.
#   We optimise the grid search directly for this.
#
# WHY StratifiedKFold?
#   Each of the 5 folds must have the same class ratio (65/35).
#   Random folds could create a fold with very few diabetic samples
#   → recall estimate becomes unreliable.
# ─────────────────────────────────────────────────────

# --- Parameter Grid ---
# n_estimators : how many trees (more = better but slower)
# max_depth    : how deep each tree can grow
#                None = unlimited (overfits); 10-20 is typical
# min_samples_split : minimum samples to split a node
# min_samples_leaf  : minimum samples in a leaf

rf_param_grid = {
    'n_estimators'     : [100, 200, 300],
    'max_depth'        : [8, 12, 15, None],
    'min_samples_split': [5, 10],
    'min_samples_leaf' : [2, 5]
}

rf_base = RandomForestClassifier(random_state=42)

# 5-fold stratified CV, scoring on recall
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

rf_grid = GridSearchCV(
    estimator=rf_base,
    param_grid=rf_param_grid,
    scoring='recall',          # optimise for recall
    cv=cv,
    n_jobs=-1,                 # use all CPU cores (parallel)
    refit=True,                # retrain best model on full train set
    return_train_score=True    # also record train scores (for overfitting check)
)

print(f'Running GridSearchCV... ({len(rf_param_grid["n_estimators"]) * len(rf_param_grid["max_depth"]) * len(rf_param_grid["min_samples_split"]) * len(rf_param_grid["min_samples_leaf"])} parameter combinations × 5 folds)')

rf_grid.fit(X_train, y_train)  # trees don't need scaling

print(f'\n✅ Best parameters found:')
print(json.dumps(rf_grid.best_params_, indent=2))
print(f'\nBest CV Recall: {rf_grid.best_score_:.4f}')
```

```python
# ─────────────────────────────────────────────────────
# EVALUATE the best Random Forest on the held-out test set
# ─────────────────────────────────────────────────────

rf_best = rf_grid.best_estimator_

rf_pred = rf_best.predict(X_test)
rf_prob = rf_best.predict_proba(X_test)[:, 1]

rf_results = evaluate_model('Random Forest', y_test, rf_pred, rf_prob)
all_results.append(rf_results)

print('=== Random Forest (Tuned) — Classification Report ===')
print(classification_report(y_test, rf_pred, target_names=['Non-Diabetic', 'Diabetic']))

print(f"Recall:    {rf_results['Recall']:.4f}")
print(f"F1-Score:  {rf_results['F1-Score']:.4f}")
print(f"MCC (Phi): {rf_results['MCC (Phi)']:.4f}")
print(f"ROC-AUC:   {rf_results['ROC-AUC']:.4f}")
print(f"Accuracy:  {rf_results['Accuracy']:.4f}")
```

```python
# Confusion matrix + Feature Importance side by side
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- Left: Confusion Matrix ---
cm_rf = confusion_matrix(y_test, rf_pred)
sns.heatmap(cm_rf, annot=True, fmt='d', cmap='Purples',
            xticklabels=['Non-Diabetic', 'Diabetic'],
            yticklabels=['Non-Diabetic', 'Diabetic'],
            ax=axes[0], linewidths=1, linecolor='black', cbar=False)
axes[0].set_xlabel('Predicted')
axes[0].set_ylabel('Actual')
axes[0].set_title('Random Forest — Confusion Matrix')

# --- Right: Feature Importance ---
# WHY feature importance matters:
#   Shows which features the forest actually used to split.
#   If engineered features rank high → they added value.
#   If original features still dominate → engineering was neutral.

feat_names = X.columns
importances = rf_best.feature_importances_
sorted_idx = np.argsort(importances)

axes[1].barh(feat_names[sorted_idx], importances[sorted_idx],
             color='#8e44ad', edgecolor='black', height=0.6)
axes[1].set_xlabel('Feature Importance (Gini)')
axes[1].set_title('Random Forest — Feature Importance')

plt.tight_layout()
plt.show()

print(f"  FN (diabetic missed): {cm_rf[1][0]}  | TP (diabetic caught): {cm_rf[1][1]}")
```

```python
# ─────────────────────────────────────────────────────
# XGBOOST — Extreme Gradient Boosting
#
# What it does (different from Random Forest!):
#   1. Train tree #1 on the original data
#   2. Compute the ERRORS (residuals) of tree #1
#   3. Train tree #2 to predict those errors
#   4. Compute errors of (tree #1 + tree #2)
#   5. Train tree #3 to predict those new errors
#   ... repeat N times
#   Final prediction = sum of all trees' predictions
#
# Key difference from Random Forest:
#   RF: trees are INDEPENDENT (parallel, vote)
#   XGB: trees are SEQUENTIAL (each fixes the last's mistakes)
#   → XGB typically achieves higher accuracy on tabular data
#
# Hyperparameters explained:
#   n_estimators   → number of boosting rounds (trees)
#   max_depth      → depth of each individual tree
#                    shallow trees = weaker learners = less overfitting
#   learning_rate  → how much each tree contributes
#                    low = each tree matters less = more robust
#                    high = faster learning but overfits faster
#   subsample      → fraction of rows used per tree (like RF bootstrap)
#   colsample_bytree → fraction of features used per tree
#   scale_pos_weight → upweights the positive class (diabetic)
#                      = ratio of negatives to positives = 500/268 ≈ 1.87
#                      This tells XGBoost "missing a diabetic case
#                      costs 1.87× more than a false alarm"
# ─────────────────────────────────────────────────────

# Class imbalance ratio for scale_pos_weight
neg_count = (y_train == 0).sum()
pos_count = (y_train == 1).sum()
scale_pos = neg_count / pos_count
print(f'scale_pos_weight = {neg_count}/{pos_count} = {scale_pos:.2f}')

xgb_param_grid = {
    'n_estimators'      : [100, 200, 300],
    'max_depth'         : [3, 5, 7],
    'learning_rate'     : [0.05, 0.1, 0.2],
    'subsample'         : [0.8],
    'colsample_bytree'  : [0.8]
}

xgb_base = XGBClassifier(
    scale_pos_weight=scale_pos,   # handle class imbalance
    random_state=42,
    eval_metric='logloss',        # required by newer XGBoost
    use_label_encoder=False
)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

xgb_grid = GridSearchCV(
    estimator=xgb_base,
    param_grid=xgb_param_grid,
    scoring='recall',
    cv=cv,
    n_jobs=-1,
    refit=True,
    return_train_score=True
)

n_combos = 1
for v in xgb_param_grid.values():
    n_combos *= len(v)
print(f'\nRunning GridSearchCV... ({n_combos} combinations × 5 folds)')

xgb_grid.fit(X_train, y_train)

print(f'\n✅ Best parameters found:')
print(json.dumps(xgb_grid.best_params_, indent=2))
print(f'\nBest CV Recall: {xgb_grid.best_score_:.4f}')
```

```python
# ─────────────────────────────────────────────────────
# EVALUATE the best XGBoost on the held-out test set
# ─────────────────────────────────────────────────────

xgb_best = xgb_grid.best_estimator_

xgb_pred = xgb_best.predict(X_test)
xgb_prob = xgb_best.predict_proba(X_test)[:, 1]

xgb_results = evaluate_model('XGBoost', y_test, xgb_pred, xgb_prob)
all_results.append(xgb_results)

print('=== XGBoost (Tuned) — Classification Report ===')
print(classification_report(y_test, xgb_pred, target_names=['Non-Diabetic', 'Diabetic']))

print(f"Recall:    {xgb_results['Recall']:.4f}")
print(f"F1-Score:  {xgb_results['F1-Score']:.4f}")
print(f"MCC (Phi): {xgb_results['MCC (Phi)']:.4f}")
print(f"ROC-AUC:   {xgb_results['ROC-AUC']:.4f}")
print(f"Accuracy:  {xgb_results['Accuracy']:.4f}")
```

```python
# Confusion matrix + Feature Importance for XGBoost
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- Left: Confusion Matrix ---
cm_xgb = confusion_matrix(y_test, xgb_pred)
sns.heatmap(cm_xgb, annot=True, fmt='d', cmap='YlOrRd',
            xticklabels=['Non-Diabetic', 'Diabetic'],
            yticklabels=['Non-Diabetic', 'Diabetic'],
            ax=axes[0], linewidths=1, linecolor='black', cbar=False)
axes[0].set_xlabel('Predicted')
axes[0].set_ylabel('Actual')
axes[0].set_title('XGBoost — Confusion Matrix')

# --- Right: Feature Importance ---
xgb_importances = xgb_best.feature_importances_
sorted_idx_xgb  = np.argsort(xgb_importances)

axes[1].barh(feat_names[sorted_idx_xgb], xgb_importances[sorted_idx_xgb],
             color='#e67e22', edgecolor='black', height=0.6)
axes[1].set_xlabel('Feature Importance')
axes[1].set_title('XGBoost — Feature Importance')

plt.tight_layout()
plt.show()

print(f"  FN (diabetic missed): {cm_xgb[1][0]}  | TP (diabetic caught): {cm_xgb[1][1]}")
```

```python
# ─────────────────────────────────────────────────────
# Compare Random Forest vs XGBoost head-to-head
# ─────────────────────────────────────────────────────

adv_df = pd.DataFrame(all_results).set_index('Model')
priority_order = ['Recall', 'F1-Score', 'MCC (Phi)', 'ROC-AUC', 'Precision', 'Accuracy']
adv_df = adv_df[priority_order]

print('=== ADVANCED MODEL COMPARISON ===')
print(adv_df.round(4).to_string())

# Determine the winner per metric
print('\n--- Winner per metric ---')
for metric in priority_order:
    winner = adv_df[metric].idxmax()
    print(f'  {metric:<12} → {winner} ({adv_df[metric].max():.4f})')
```

```python
# ─────────────────────────────────────────────────────
# ROC Curves — Advanced Models only
# ─────────────────────────────────────────────────────

fig, ax = plt.subplots(figsize=(8, 7))

for name, probs, color in [
    ('Random Forest', rf_prob, '#8e44ad'),
    ('XGBoost',       xgb_prob, '#e67e22')
]:
    fpr, tpr, _ = roc_curve(y_test, probs)
    auc = roc_auc_score(y_test, probs)
    ax.plot(fpr, tpr, color=color, linewidth=3, label=f'{name} (AUC = {auc:.3f})')

ax.plot([0, 1], [0, 1], 'k--', linewidth=1, label='Random (AUC = 0.5)')

ax.set_xlabel('False Positive Rate')
ax.set_ylabel('True Positive Rate (Recall)')
ax.set_title('ROC Curves — Advanced Models')
ax.legend(loc='lower right', fontsize=12)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# ─────────────────────────────────────────────────────
# MODEL SELECTION LOGIC:
#
# We DO NOT pick the model with the highest accuracy.
# We pick the model with the best RECALL first.
# If recall is tied, we break the tie with F1-Score.
# If still tied, MCC decides.
#
# WHY this priority?
#   A diabetic patient who is NOT caught by the model
#   may go undiagnosed for years → organ damage, blindness, etc.
#   A healthy patient flagged as diabetic will simply
#   get a follow-up blood test → minor inconvenience.
#   The cost of FN >> cost of FP in medical screening.
# ─────────────────────────────────────────────────────

# Compare on recall
rf_recall  = rf_results['Recall']
xgb_recall = xgb_results['Recall']

if rf_recall > xgb_recall:
    best_model_name = 'Random Forest'
    best_model      = rf_best
    best_results    = rf_results
elif xgb_recall > rf_recall:
    best_model_name = 'XGBoost'
    best_model      = xgb_best
    best_results    = xgb_results
else:
    # Tie-break on F1
    if rf_results['F1-Score'] >= xgb_results['F1-Score']:
        best_model_name = 'Random Forest'
        best_model      = rf_best
        best_results    = rf_results
    else:
        best_model_name = 'XGBoost'
        best_model      = xgb_best
        best_results    = xgb_results

print(f'🏆 SELECTED MODEL: {best_model_name}')
print(f'   Recall:    {best_results["Recall"]:.4f}')
print(f'   F1-Score:  {best_results["F1-Score"]:.4f}')
print(f'   MCC (Phi): {best_results["MCC (Phi)"]:.4f}')
print(f'   ROC-AUC:   {best_results["ROC-AUC"]:.4f}')
print(f'   Accuracy:  {best_results["Accuracy"]:.4f}')
```

```python
# ─────────────────────────────────────────────────────
# SAVE ARTIFACTS
#
# 1. best_diabetes_model.pkl  → the trained classifier
#    The backend API loads this and calls .predict()
#
# 2. model_comparison.json    → all metrics for all models
#    Used in the evaluation notebook and README reporting
# ─────────────────────────────────────────────────────

os.makedirs('artifacts/models',  exist_ok=True)
os.makedirs('artifacts/reports', exist_ok=True)

# Save the best model
MODEL_PATH = 'artifacts/models/best_diabetes_model.pkl'
joblib.dump(best_model, MODEL_PATH)
print(f'✅ Saved best model → {MODEL_PATH}')

# Save comparison report
REPORT_PATH = 'artifacts/reports/model_comparison.json'
report = {
    'selected_model': best_model_name,
    'selection_criteria': 'Recall > F1 > MCC (medical priority)',
    'models': all_results
}
with open(REPORT_PATH, 'w') as f:
    json.dump(report, f, indent=2)
print(f'✅ Saved comparison  → {REPORT_PATH}')
```
